```python
import pandas as pd
import json
import os

import sys

# use absolute path here
project_path = "/mnt/School/PhD/AI221/Project/"
sys.path.insert(0, project_path)

from src.data_extraction.utils.constants import data_path
```

```python
fcp_path = os.path.join(data_path, "flood_control_projects")
```

```python
fcp_json_file = os.path.join(fcp_path, "flood_control_projects.json")
with open(fcp_json_file, 'r') as file:
    fcp_json = json.load(file)
    
fcp_json["features"][0]
```

```python
fcp_data = [content["attributes"] for content in fcp_json["features"]]
fcp_data[0]
```

```python
fcp_df = pd.DataFrame(fcp_data)
fcp_df["CompletionDateActual"] = pd.to_datetime(fcp_df["CompletionDateActual"]) 
fcp_df["CompletionYear"] = fcp_df.CompletionDateActual.dt.year
fcp_df.head()
```

```python
fcp_df[["CompletionYear", "CompletionDateActual"]].groupby("CompletionYear").count()
```

```python
# Validation: Should be the same as the dashboard: https://dpwh.maps.arcgis.com/apps/dashboards/f585444def084abaadda090952759e28
# Observation: Seems like the aggregates are static and not from the data (to check)
agg_df = (
    fcp_df.loc[
        fcp_df["CompletionYear"] >= 2022,
        ["CompletionYear", "Region", "Province"]
    ]
    .groupby(["CompletionYear", "Region"]).agg("count").reset_index()
)
pivot_df = agg_df.pivot(index="Region", columns="CompletionYear", values="Province")
pivot_df
```

```python
df_raw = fcp_df.copy()
outlier_reports = {}

# -----------------------------
# 1) Required columns
# -----------------------------
required_cols = ["Region", "Province", "Municipality", "CompletionDateActual"]
missing_required = [c for c in required_cols if c not in df_raw.columns]
if missing_required:
    raise ValueError(f"Missing required columns: {missing_required}")

# -----------------------------
# 2) Date parsing outliers
# -----------------------------
df_raw["CompletionDateActual_parsed"] = pd.to_datetime(
    df_raw["CompletionDateActual"], errors="coerce"
)
df_raw["CompletionYear"] = df_raw["CompletionDateActual_parsed"].dt.year

bad_dates = df_raw[df_raw["CompletionDateActual"].notna() & df_raw["CompletionDateActual_parsed"].isna()]
if len(bad_dates) > 0:
    outlier_reports["bad_dates"] = bad_dates[
        [c for c in ["CompletionDateActual", "ProjectName", "Region", "Province", "Municipality"] if c in bad_dates.columns]
    ].head(100)

out_of_range_years = df_raw[
    df_raw["CompletionYear"].notna() &
    ~df_raw["CompletionYear"].between(2022, 2025)
]
if len(out_of_range_years) > 0:
    outlier_reports["out_of_range_years"] = out_of_range_years[
        [c for c in ["CompletionDateActual", "CompletionYear", "ProjectName", "Region", "Province", "Municipality"] if c in out_of_range_years.columns]
    ].head(100)

# -----------------------------
# 3) Missing locality outliers
# -----------------------------
locality_issue_mask = (
    df_raw["Region"].isna() | df_raw["Region"].astype(str).str.strip().eq("") |
    df_raw["Province"].isna() | df_raw["Province"].astype(str).str.strip().eq("") |
    df_raw["Municipality"].isna() | df_raw["Municipality"].astype(str).str.strip().eq("")
)
bad_locality = df_raw[locality_issue_mask]
if len(bad_locality) > 0:
    outlier_reports["bad_locality"] = bad_locality[
        [c for c in ["ProjectName", "Region", "Province", "Municipality", "CompletionDateActual"] if c in bad_locality.columns]
    ].head(100)

# -----------------------------
# 4) Duplicate rows
# -----------------------------
candidate_id_cols = ["ProjectID", "project_id", "OBJECTID", "Id", "id"]
id_col = next((c for c in candidate_id_cols if c in df_raw.columns), None)

if id_col:
    dupes = df_raw[df_raw[id_col].duplicated(keep=False)].sort_values(id_col)
else:
    subset = [c for c in ["ProjectName", "Region", "Province", "Municipality", "CompletionDateActual"] if c in df_raw.columns]
    dupes = df_raw[df_raw.duplicated(subset=subset, keep=False)].sort_values(subset)

if len(dupes) > 0:
    outlier_reports["duplicates"] = dupes.head(100)

# -----------------------------
# 5) Rare / suspicious region-year buckets
# -----------------------------
region_year = (
    df_raw.dropna(subset=["Region", "CompletionYear"])
    .groupby(["Region", "CompletionYear"])
    .size()
    .reset_index(name="project_count")
)

rare_region_year = region_year[region_year["project_count"] <= 2]
if len(rare_region_year) > 0:
    outlier_reports["rare_region_year"] = rare_region_year.sort_values(["project_count", "Region", "CompletionYear"])

# -----------------------------
# 6) Save only the outliers
# -----------------------------

summary = []
for name, report_df in outlier_reports.items():
    summary.append({"check": name, "rows_flagged": len(report_df)})

summary_df = pd.DataFrame(summary)

print(summary_df if not summary_df.empty else "No major outliers found.")
```

```python
fcp_folder = "flood_control_projects"
fcp_df.to_csv(
    os.path.join(data_path, fcp_folder, "flood_control_raw.csv"),
    index=False
)
```
