```python
import os
import geopandas as gpd
import numpy as np
import pandas as pd

import sys

# use absolute path here
project_path = "/mnt/School/PhD/AI221/Project/"
sys.path.insert(0, project_path)

from src.data_extraction.utils.constants import data_path
```

```python
urban_datasets_dict = {
    "active_flood_days": "urban_active_flood_days.csv",
    "annual_nightlights": "urban_annual_nightlights.csv",
    "flood_control_projects": "urban_flood_control_projects.csv",
    "monthly_rainfall": "urban_monthly_rainfall.csv",
    "osm_metrics": "urban_osm_metrics.csv",
    "twi_metrics": "urban_twi_metrics.csv",
}
```

### Setting keys 

```python
psgc_join_key = "adm3_psgc"
year_month_join_key = [psgc_join_key, "year_month"]
year_join_key = [psgc_join_key, "year"]
unique_keys = year_join_key + ["year_month"]
```

### Loading datasets

```python
ph_bounds = gpd.read_file(os.path.join(data_path,"ph_adm3_municities/PH_Adm3_MuniCities.shp.shp"))
ph_bounds = ph_bounds[ph_bounds["geo_level"] == "City"]
ph_bounds.head()
```

```python
# Loading feature dataframes
active_flood_days_df = pd.read_csv(os.path.join(data_path, "urban_active_flood_days.csv"))
active_flood_days_df = active_flood_days_df.rename(
    columns={
        "psgc": "adm3_psgc",
        "month_year": "year_month"
    }
)

active_flood_days_df
```

```python
flood_control_projects_df = pd.read_csv(os.path.join(data_path, "urban_flood_control_projects.csv"))
flood_control_projects_df = flood_control_projects_df.rename(
    columns={
        colname: f"fcp_{colname}"
        for colname in flood_control_projects_df.columns
        if colname not in year_month_join_key
    }
)
flood_control_projects_df.head()
```

```python
osm_metrics_df = pd.read_csv(os.path.join(data_path, "urban_osm_metrics.csv"))
osm_metrics_df = osm_metrics_df.drop(columns=["city_name"])
osm_metrics_df = osm_metrics_df .rename(
    columns={
        "psgc": "adm3_psgc"
    }
)
osm_metrics_df = osm_metrics_df.rename(
    columns={
        colname: f"osm_{colname}"
        for colname in osm_metrics_df.columns
        if colname not in year_month_join_key + ["year"]
    }
)

osm_metrics_df.head()
```

```python
annual_nightlights_df = pd.read_csv(os.path.join(data_path, "urban_annual_nightlights.csv"))
annual_nightlights_df = annual_nightlights_df .rename(
    columns={
        "psgc": "adm3_psgc"
    }
)
annual_nightlights_df = annual_nightlights_df .rename(
    columns={
        colname: "nightlight_" + colname.replace("_nightlight", "")
        for colname in annual_nightlights_df.columns
        if colname not in year_month_join_key + ["year"]
    }
)
annual_nightlights_df 
```

```python
twi_metrics_df = pd.read_csv(os.path.join(data_path, "urban_twi_metrics.csv"))
twi_metrics_df  = twi_metrics_df.drop(columns=["adm3_en"])
twi_metrics_df 
```

```python
monthly_rainfall_df = pd.read_csv(os.path.join(data_path, "urban_monthly_rainfall.csv"))
monthly_rainfall_df 
```

## Consolidating data

```python
# Constant; one value for all year-months
consolidated_df_twi = pd.merge(
    ph_bounds[["adm3_psgc", "adm3_en"]],
    twi_metrics_df,
    on=psgc_join_key,
    how="left"
)
consolidated_df_twi.shape
```

```python
# Builds the year-month permutations
consolidated_df_rain = pd.merge(
    consolidated_df_twi,
    monthly_rainfall_df,
    on=psgc_join_key,
    how="left"
)
consolidated_df_rain = consolidated_df_rain.rename(
    columns={
        "date": "year_month",
        "avg": "monthly_avg_rainfall",
    }
)
consolidated_df_rain["year"] = consolidated_df_rain["year_month"].apply(
    lambda x: int(x.split("-")[0])
)
consolidated_df_rain.shape
```

```python
# Annual data
consolidated_df_osm = pd.merge(
    consolidated_df_rain,
    osm_metrics_df,
    on=year_join_key,
    how="outer"
)
consolidated_df_osm.shape
```

```python
# Monthly data
consolidated_df_fcp = pd.merge(
    consolidated_df_osm,
    flood_control_projects_df,
    on=year_month_join_key,
    how="outer"
)
consolidated_df_fcp.shape
```

```python
# Annual data
consolidated_dataframe_nightlights = pd.merge(
    consolidated_df_fcp,
    annual_nightlights_df,
    on=year_join_key,
    how="outer"
)
consolidated_dataframe_nightlights.shape
```

```python
# Monthly data
final_consolidated_dataframe_nightlights = pd.merge(
    consolidated_dataframe_nightlights,
    active_flood_days_df,
    on=year_month_join_key,
    how="outer"
).drop_duplicates(subset=unique_keys)
final_consolidated_dataframe_nightlights 
```
